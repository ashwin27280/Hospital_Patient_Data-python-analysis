
2. Select only the columns relevant for billing: ['PatientID', 'Department', 'Doctor', 'BillAmount']

```python
bdf.columns
```

```python
# Select relevant columns for billing
billing_relevant=bdf[['PatientID', 'Department', 'Doctor', 'BillAmount']]
```

```python
#show the result
print("Relevant billing columns: \n")
billing_relevant.head()
```


3. Drop administrative columns like ['ReceptionistID', 'CheckInTime']

```python
bdf.head()
```

```python
bdf=bdf.drop(columns=['ReceptionistID', 'CheckInTime'], errors='ignore')
```

```python
#after droppin the adminstrative columns
bdf.head()
```

4. Use groupby to find total bill amount per department


```python
dept_bill_summary=billing_relevant.groupby('Department')['BillAmount'].sum().reset_index()
```

```python
dept_bill_summary
#total bill amount per department
```

5. Remove duplicate patient records based on PatientID

```python
bdf.duplicated().sum()
```

```python
#remove duplicate patient records based on patientID
patient_unique = bdf.drop_duplicates(subset=['PatientID'],inplace=True)
```

```python
# After removing duplicate PatientID records
patient_unique.head()
```

```python
bdf.duplicated().sum()
```


6. Fill missing BillAmount values with the mean bill amount

```python
# Fill missing BillAmount values with the mean bill amount
mean_bill=patient_unique['BillAmount'].mean()

patient_unique['BillAmount'].fillna(mean_bill,inplace=True)
```

```python
patient_unique
```

```python
patient_unique.isnull().sum()
```

7. Merge the billing dataset with patient dataset on PatientID


```python
merge_df=pd.merge(left=bdf,right=pdf,how='inner',on='PatientID')
```

```python
merge_df
```

8. Concatenate an additional DataFrame that contains new patients for the current week (row-wise)

```python
# Concatenate new patients (row-wise)
new_patients = pd.DataFrame({
    'PatientID':[107,108],
    'Name':['Martin','lewis'],
    'Department':['oncology','cardiology'],
    'Doctor':['Dr.Ashwin','Dr.Piyush'],
    'BillAmount':[9000,6500],
    'InsuranceCovered':[3000,2500],
    'FinalAmount':[5000,3500]

})
```

```python
# Concatenate with existing merged data
full_patientdf=pd.concat([bdf,new_patients],ignore_index=True)
```

```python
full_patientdf
```

9. Concatenate new billing category columns like ['InsuranceCovered', 'FinalAmount'] (column-wise)

```python
# new billing column
new_billing_column=pd.DataFrame({
    'InsuranceCovered':[3000,2000,1000,5000,4000,2500,None,None],
    'FinalAmount':[5000,3000,1000,6000,5500,3500,None,None]
})
```

```python
# Concatenate new billing category (column-wise)
final_df = pd.concat([merge_df.reset_index(drop=True), new_billing_column], axis=1)
merge_df['FinalAmount'] = merge_df['BillAmount'] - merge_df['InsuranceCovered']
```
